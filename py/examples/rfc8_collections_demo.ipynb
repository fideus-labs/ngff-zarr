# RFC-8 on a real multimodal case

Imaging data rarely comes as one image. A patient has a CT and an MR, each has
a segmentation, and a registration relates them. Nothing in OME-Zarr said how
to write that down, so every tool invented its own way, and
[RFC-8](https://ngff.openmicroscopy.org/rfc/8/index.html) quotes the result:
users maintain tool-specific metadata for the same relationships, and archives
publish the same collection several times over for different consumers.

What it changes, concretely, is that these four things become one document.
The segmentation stops being bound to its image by sitting inside its folder
and says which image it annotates. The registration stops being a file beside
the data and becomes a transformation the document declares. And a member can
live anywhere, so the images below stay in the archive that published them
while the derived products sit on an object store.

Three constructs do that work.

A **node** is any thing the format can name: an image, a label map, a group of
them. It carries a type, a name, and its own attributes.

A **path** locates a node somewhere else. It is `{type, path}`, where the path
is relative to the document, an absolute `file://` path, or an `http(s)` URL.

A **reference** points at another node by `id`, adding a `path` when that node
lives in another document. This is what lets a label map say which image it
annotates without having to sit inside it.

The case below is one patient from the SynthRAD2025 challenge: an abdominal CT
and MR, each segmented, plus the deformable registration between them. The four
pieces live in three different places, and RFC-8 is what holds them together.


```python
import io
import json
import urllib.request
import zipfile
from pathlib import Path

import matplotlib.pyplot as plt
import ngff_zarr as nz
import numpy as np
from ngff_zarr import rfc8

BASE = "https://ome-zarr-rfc8.s3.filebase.io/synthrad-1ABB070"
ARCHIVE = (
    "https://zenodo.org/api/records/15373853/files/synthRAD2025_Task1_Train.zip/content"
)
CASE = "synthRAD2025_Task1_Train/Task1/AB/1ABB070"
WORK = Path("rfc8-demo-work")
WORK.mkdir(exist_ok=True)

print("ngff-zarr", nz.__version__)
```

## The images stay where they were published

The CT and the MR are not copied anywhere. They sit inside a 14.8 GB archive on
Zenodo, and Zenodo answers range requests, so the two files come out of it
without downloading the rest: read the zip's central directory, then fetch one
member's compressed bytes in a single request.


```python
class RemoteFile(io.RawIOBase):
    """A read-only file over HTTP, backed by range requests."""

    def __init__(self, url):
        self.url, self.pos = url, 0
        request = urllib.request.Request(url, method="HEAD")
        with urllib.request.urlopen(request) as response:
            self.size = int(response.headers["Content-Length"])

    def seekable(self):
        return True

    def readable(self):
        return True

    def seek(self, offset, whence=0):
        self.pos = (offset, self.pos + offset, self.size + offset)[whence]
        return self.pos

    def tell(self):
        return self.pos

    def readinto(self, buffer):
        data = self.read(len(buffer))
        buffer[: len(data)] = data
        return len(data)

    def read(self, count=-1):
        if count is None or count < 0:
            count = self.size - self.pos
        end = min(self.pos + count, self.size) - 1
        request = urllib.request.Request(
            self.url, headers={"Range": f"bytes={self.pos}-{end}"}
        )
        with urllib.request.urlopen(request) as response:
            data = response.read()
        self.pos += len(data)
        return data


archive = zipfile.ZipFile(io.BufferedReader(RemoteFile(ARCHIVE), 1 << 20))
for name in ("ct.mha", "mr.mha"):
    target = WORK / name
    if not target.exists():
        with archive.open(f"{CASE}/{name}") as member:
            target.write_bytes(member.read())
    print(
        f"{name}: {target.stat().st_size / 1e6:.1f} MB out of {archive.fp.raw.size / 1e9:.1f} GB"
    )
```

Converting them to OME-Zarr is one call each. Written at `0.9.dev3`, the root
`ome` value of each store *is* a multiscale node, which is the change RFC-8
makes to the core model.


```python
stores = {}
for name in ("ct", "mr"):
    backend = nz.detect_cli_io_backend([str(WORK / f"{name}.mha")])
    image = nz.cli_input_to_ngff_image(backend, [str(WORK / f"{name}.mha")])
    image.name = name
    store = (WORK / f"{name}.ome.zarr").resolve()
    nz.to_ome_zarr(
        str(store),
        nz.to_multiscales(image, chunks=64),
        version="0.9.dev3",
        overwrite=True,
    )
    stores[name] = store
    print(
        f"{name}: read with {backend.value}, {image.data.shape} at "
        f"{tuple(round(v, 1) for v in image.scale.values())} mm -> {store.name}"
    )

root = json.loads((stores["ct"] / "zarr.json").read_text())["attributes"]["ome"]
print("\nthe store's own root node:", {k: root[k] for k in ("version", "type", "name")})
```

```python
ct = nz.from_ome_zarr(str(stores["ct"])).images[0]
mr = nz.from_ome_zarr(str(stores["mr"])).images[0]

body = np.asarray(ct.data[ct.data.shape[0] // 2]) > -500
rows, cols = np.where(body)
crop = (slice(rows.min(), rows.max()), slice(cols.min(), cols.max()))
SLICE = 70

figure, axes = plt.subplots(1, 2, figsize=(11, 5.2))
for axis, image, title, window in (
    (axes[0], ct, "CT", (-160, 240)),
    (axes[1], mr, "MR", (0, 600)),
):
    plane = np.asarray(image.data[SLICE])[crop]
    axis.imshow(
        np.clip((plane - window[0]) / (window[1] - window[0]), 0, 1), cmap="gray"
    )
    axis.set_title(f"{title}, slice {SLICE}")
    axis.set_xticks([])
    axis.set_yticks([])
figure.suptitle("SynthRAD2025 Task 1 abdomen, case 1ABB070, read from Zenodo")
figure.tight_layout()
```

## The segmentations carry their own label metadata

TotalSegmentator ran on each image through KonfAI, and the two label maps were
published on their own, separate from the images. Each is an RFC-8 multiscale
node whose `labels` attribute travels with it.

Under RFC-8 that attribute holds one `labelAttributes` entry per value, with
`labelValue`, an RGBA `color`, and whatever else the producer wants to keep. The
v0.4 `image-label` metadata split that across two parallel lists, `colors` and
`properties`, joined by a repeated `label-value` field; nothing forced them to
agree, or to be the same length. Here the structure name comes back under a
prefixed key, which is how RFC-8 says an extension carries its own data.


```python
labels = {}
for side in ("ct", "mr"):
    _, document = rfc8.resolve_ome_document(
        rfc8.OmePath(
            type=rfc8.PATH_TYPE_ZARR, path=f"{BASE}/{side}-totalsegmentator.ome.zarr"
        )
    )
    node, version = rfc8.node_from_ome_dict(document["ome"])
    labels[side] = rfc8.labels(node)
    print(
        f"{side}: {node.type} node at version {version}, "
        f"{len(labels[side].labelAttributes)} label entries"
    )

for entry in labels["ct"].labelAttributes[:4]:
    print(f"  labelValue {entry.labelValue:3d}  color {entry.color}  {entry.extra}")
```

```python
from matplotlib.colors import ListedColormap


def label_colormap(entries):
    """A colormap straight out of the RFC-8 label attributes."""
    size = int(max(entry.labelValue for entry in entries)) + 1
    colors = np.zeros((size, 4))
    for entry in entries:
        colors[int(entry.labelValue)] = [channel / 255 for channel in entry.color]
    colors[0] = [0, 0, 0, 0]
    return ListedColormap(colors)


segmentations = {
    side: nz.from_ome_zarr(f"{BASE}/{side}-totalsegmentator.ome.zarr").images[0]
    for side in ("ct", "mr")
}

figure, axes = plt.subplots(1, 2, figsize=(11, 5.4))
for axis, side, image, window in (
    (axes[0], "ct", ct, (-160, 240)),
    (axes[1], "mr", mr, (0, 600)),
):
    plane = np.asarray(image.data[SLICE])[crop]
    labelmap = np.asarray(segmentations[side].data[SLICE])[crop]
    colormap = label_colormap(labels[side].labelAttributes)
    axis.imshow(
        np.clip((plane - window[0]) / (window[1] - window[0]), 0, 1), cmap="gray"
    )
    axis.imshow(
        np.ma.masked_where(labelmap == 0, labelmap),
        cmap=colormap,
        vmin=0,
        vmax=colormap.N - 1,
        interpolation="nearest",
        alpha=0.55,
    )
    axis.set_title(
        f"{side.upper()}: {len(labels[side].labelAttributes)} structures, "
        f"{len(np.unique(labelmap)) - 1} on this slice"
    )
    axis.set_xticks([])
    axis.set_yticks([])
figure.suptitle("Every colour comes from the labelAttributes of the store itself")
figure.tight_layout()
```

No palette was chosen here: each colour is the `color` of the matching
`labelValue`, read out of the node the object store holds. Under v0.4 that
would have taken the `colors` list and the `properties` list side by side,
matched on a repeated `label-value` with nothing to keep them in step.

## One collection, three locations

The images are on this machine, the label maps and the registration are on an
object store. A collection node holds all four, each located by its own path,
and the two label nodes name the image they annotate through a `source`
reference rather than by sitting inside it.

That reference is what makes the arrangement expressible at all. In the v0.4
layout a segmentation was bound to its image by being stored in its `labels/`
subgroup, so the relationship *was* the location. Here it survives the two
living apart, and nothing stops a segmentation from naming several sources, or
several segmentations from naming one.


```python
from ngff_zarr.v06.zarr_metadata import (
    Axis,
    CoordinateSystem,
    CoordinateSystemIdentifier,
    Displacements,
)


def node(node_id, name, path):
    return rfc8.Node(
        type="multiscale",
        id=node_id,
        name=name,
        path=rfc8.OmePath(type=rfc8.PATH_TYPE_ZARR, path=path),
    )


def label_node(node_id, name, store, source):
    value = node(node_id, name, f"{BASE}/{store}")
    rfc8.set_labels(value, rfc8.Labels(source=[rfc8.Reference(id=source)]))
    return value


collection = rfc8.Collection(
    id="synthrad-1ABB070",
    name="SynthRAD2025 Task 1 abdomen, case 1ABB070",
    nodes=[
        node("ct", "planning CT", stores["ct"].as_uri()),
        node("mr", "abdominal MR", stores["mr"].as_uri()),
        label_node(
            "ct-labels",
            "TotalSegmentator on the CT",
            "ct-totalsegmentator.ome.zarr",
            "ct",
        ),
        label_node(
            "mr-labels",
            "TotalSegmentator on the MR",
            "mr-totalsegmentator.ome.zarr",
            "mr",
        ),
    ],
)

axes = [Axis(name=n, type="space", unit="millimeter") for n in ("z", "y", "x")]
rfc8.set_scene(
    collection,
    rfc8.Scene(
        coordinateSystems=[
            CoordinateSystem(id="ct-space", name="ct", axes=axes),
            CoordinateSystem(id="mr-space", name="mr", axes=axes),
        ],
        coordinateTransformations=[
            Displacements(
                name="IMPACT B-spline, MR onto CT",
                path=f"{BASE}/mr-to-ct-displacements.ome.zarr",
                input=CoordinateSystemIdentifier(id="ct-space"),
                output=CoordinateSystemIdentifier(id="mr-space"),
            )
        ],
    ),
)

rfc8.validate_collection(collection)
for value in collection.nodes:
    source = rfc8.labels(value)
    annotates = f" annotates {source.source[0].id}" if source and source.source else ""
    where = "local" if value.path.path.startswith("file://") else "object store"
    print(f"  {value.id:10s} {where:12s} {value.name}{annotates}")
```

```python
document = rfc8.to_collection_json(collection)
back, declared = rfc8.node_from_ome_dict(document["ome"])
print("declared version:", declared)
round_trip_intact = [n.id for n in back.nodes] == [n.id for n in collection.nodes]
print("round trip preserves the nodes:", round_trip_intact)
print()
print(
    json.dumps(
        {k: document["ome"][k] for k in ("version", "type", "id", "name")}, indent=1
    )
)
print("scene transformation:")
print(
    json.dumps(
        document["ome"]["attributes"]["scene"]["coordinateTransformations"][0], indent=1
    )
)
```

## What the rules refuse

The collection above validates. These do not, and each failure names the rule
it broke and where.


```python
def rejected(name, mutate):
    broken = json.loads(json.dumps(document))
    mutate(broken["ome"])
    try:
        node_value, _ = rfc8.node_from_ome_dict(broken["ome"])
        rfc8.validate_collection(node_value)
    except Exception as error:
        rule = getattr(error, "rule", type(error).__name__)
        where = getattr(error, "location", "")
        print(f"{name}\n  {rule} at {where}\n  {error}\n")
    else:
        print(f"{name}: accepted, which it should not be\n")


rejected(
    "two nodes share a name",
    lambda ome: ome["nodes"][1].update(name=ome["nodes"][0]["name"]),
)
rejected(
    "an id with a space in it", lambda ome: ome["nodes"][0].update(id="planning ct")
)
rejected(
    "a source reference to an id this document does not declare",
    lambda ome: ome["nodes"][2]["attributes"]["labels"].update(
        source=[{"id": "elsewhere"}]
    ),
)
rejected(
    "a path type nothing defines",
    lambda ome: ome["nodes"][0]["path"].update(type="zip"),
)
```

The third one is the interesting refusal. A reference whose `id` is not declared
in the document is external, and RFC-8 requires an external reference to say
where that document is. Without the rule a collection could quietly point at
nothing.


## The registration is declared, not baked in

Neither label map was resampled before being published. Each sits in the space
it was computed in, and the collection declares how to get from one to the
other. That is deliberate: warping a segmentation before storing it hides the
relationship in the voxels, where it cannot be inspected, inverted, or replaced.

`nz.resample` takes the transformation straight off the scene. The field is a
100 x 439 x 442 x 3 store on the object store, and it is streamed: each output
block reads only the window of the field its own points fall in, then only the
region of the moving image that window reaches. Neither has to fit in memory.


```python
scene = rfc8.scene(collection)
transform = scene.coordinateTransformations[0]
field = nz.from_ome_zarr(transform.path)
print(
    f"{transform.type} field {field.images[0].data.shape} {field.images[0].data.dtype}"
)

ct_labels = nz.from_ome_zarr(f"{BASE}/ct-totalsegmentator.ome.zarr").images[0]
mr_labels = nz.from_ome_zarr(f"{BASE}/mr-totalsegmentator.ome.zarr").images[0]

moved = nz.resample(
    transform,
    fixed=ct_labels,
    moving=mr_labels,
    interpolator="label_image",
    fields={transform.path: field},
)
print("resampled lazily:", moved.data.shape, moved.data.dtype)
```

Whether that transformation says something true is checkable. Both label maps
name their structures, so the organs they share can be compared before and after
it is applied.


```python
names = {}
for side, value in labels.items():
    names[side] = {
        entry.extra.get("konfai:name"): int(entry.labelValue)
        for entry in value.labelAttributes
    }

CORE = [
    "liver",
    "spleen",
    "kidney_left",
    "kidney_right",
    "stomach",
    "pancreas",
    "aorta",
]
reference = np.asarray(ct_labels.data)
before = np.asarray(mr_labels.data)
after = np.asarray(moved.data)


def dice(a, b):
    total = a.sum() + b.sum()
    return 2.0 * np.logical_and(a, b).sum() / total if total else float("nan")


print(f"{'organ':14s} {'before':>7s} {'after':>7s}")
scores = []
for organ in CORE:
    target = reference == names["ct"][organ]
    if target.sum() < 500:
        continue
    b = dice(target, before == names["mr"][organ])
    a = dice(target, after == names["mr"][organ])
    scores.append((b, a))
    print(f"{organ:14s} {b:7.3f} {a:7.3f}")
print(
    f"{'mean':14s} {np.mean([s[0] for s in scores]):7.3f} {np.mean([s[1] for s in scores]):7.3f}"
)
```

```python
background = np.clip((np.asarray(ct.data[SLICE])[crop] + 160) / 400, 0, 1)

figure, axes = plt.subplots(1, 2, figsize=(11, 5.4))
for axis, labelmap, title in (
    (axes[0], before[SLICE], "before the transformation"),
    (axes[1], after[SLICE], "after it"),
):
    axis.imshow(background, cmap="gray")
    for organ in CORE:
        outline = (reference[SLICE] == names["ct"][organ])[crop]
        if outline.any():
            axis.contour(outline, levels=[0.5], colors="#ffd400", linewidths=1.0)
        moving = (labelmap == names["mr"][organ])[crop]
        if moving.any():
            axis.contour(moving, levels=[0.5], colors="#ff2d55", linewidths=1.0)
    axis.set_title(title)
    axis.set_xticks([])
    axis.set_yticks([])
figure.suptitle("yellow: organs segmented on the CT.  red: organs segmented on the MR")
figure.tight_layout()
```

## One document, opened by anything

The reason RFC-8 exists is that every tool grew its own way of grouping images,
so an archive ends up publishing the same collection several times over. What
replaces that is a single document, and the store above carries one:
`collection.json` beside the three arrays it names.

Its members are named by paths relative to it, so the whole prefix can be moved
or mirrored without rewriting anything. Opening it is one call, and the relative
paths resolve against wherever the document turned out to be.


```python
derived = rfc8.from_collection_json(f"{BASE}/collection.json")
print("id:", derived.root.id)
print("base:", derived.base)
for value in derived.root.nodes:
    print(f"  {value.id:10s} {value.path.type:5s} {value.path.path}")
```

A collection node holds either its nodes inline or a path to them, never both.
The study below takes the second form: the two images sit beside the document
as relative paths, and the derived products are one member reached by a `json`
path across the network. Nothing is fetched while the document is being read.


```python
study_nodes = [
    rfc8.Node(
        type="multiscale",
        id="ct",
        name="planning CT",
        path=rfc8.OmePath(type=rfc8.PATH_TYPE_ZARR, path="./ct.ome.zarr"),
    ),
    rfc8.Node(
        type="multiscale",
        id="mr",
        name="abdominal MR",
        path=rfc8.OmePath(type=rfc8.PATH_TYPE_ZARR, path="./mr.ome.zarr"),
    ),
    rfc8.Collection(
        id="derived",
        name="Segmentations and registration",
        path=rfc8.OmePath(type=rfc8.PATH_TYPE_JSON, path=f"{BASE}/collection.json"),
    ),
]
study = rfc8.Collection(id="study", name="SynthRAD2025 case 1ABB070", nodes=study_nodes)
rfc8.set_scene(study, rfc8.scene(collection))
rfc8.validate_collection(study)

rfc8.to_collection_json(study, path=str(WORK / "study.json"))
reopened = rfc8.from_collection_json(str(WORK / "study.json"))
print("opened", (WORK / "study.json").name, "with base", Path(reopened.base).name)
for value in reopened.root.nodes:
    holds = "nodes inline" if value.nodes else f"path -> {value.path.path}"
    print(f"  {value.type:11s} {value.id:8s} {holds}")
```

Following the nested member fetches the other document and hands back a
collection of its own, anchored at the location it came from. Its members
resolve from there, so the arrays it names are reachable without the study
document knowing anything about them.


```python
nested = reopened.load("derived")
print(type(nested).__name__, "at", nested.base)
for value in nested.root.nodes:
    print(f"  {value.id}")

image = nested.load("ct-labels")
print(
    "\nloaded through two documents:",
    type(image).__name__,
    image.images[0].data.shape,
    image.images[0].data.dtype,
)
```

A reference names a node by `id`. Inside one document that is all it needs; the
`source` links above are of that kind. Pointing into another document takes a
`path` as well, and the rule that refuses a pathless external reference is the
one that fired earlier.


```python
external = rfc8.Reference(
    id="mr-labels",
    path=rfc8.OmePath(type=rfc8.PATH_TYPE_JSON, path=f"{BASE}/collection.json"),
)
resolved = reopened.load(external)
print(
    f"reference {external.id!r} in {external.path.path.rsplit('/', 1)[-1]}"
    f" -> {type(resolved).__name__} {resolved.images[0].data.shape}"
)
```

## Extensibility

The other half of the RFC is that a reader meets things it does not know. A
prefixed identifier marks them, at every point the specification names: node
types, attribute keys, path types, transformation types, axis types. Unprefixed
values belong to the core specification alone.

`konfai:name` above was one of those, on a label entry. A whole node can be
foreign in the same way, and a reader that does not recognise the type is meant
to carry it along rather than choke on it.


```python
study_nodes.append(
    rfc8.Node(
        type="acme:mesh",
        id="liver-surface",
        name="Liver surface mesh",
        path=rfc8.OmePath(type="acme:ply", path="./liver.ply"),
        attributes={"acme:vertices": 20344},
    )
)
extended = rfc8.Collection(id="study", name=study.name, nodes=study_nodes)
rfc8.validate_collection(extended)

document = rfc8.to_collection_json(extended)
back, _ = rfc8.node_from_ome_dict(document["ome"])
foreign = back.nodes[-1]
print(
    f"{foreign.type} node survived the round trip: {foreign.name!r},"
    f" path type {foreign.path.type}, attributes {foreign.attributes}"
)
```

## What ran

Every claim above is a line of output, not a description. Collected:


```python
checks = [
    ("the root ome value is a node", root["type"] == "multiscale"),
    (
        "a node carries its own label attributes",
        len(labels["ct"].labelAttributes) == 76,
    ),
    (
        "a label map names the image it annotates",
        rfc8.labels(collection.nodes[2]).source[0].id == "ct",
    ),
    (
        "a collection holds members in two locations",
        {n.path.path.split(":")[0] for n in collection.nodes} == {"file", "https"},
    ),
    (
        "all three path forms appear",
        {"relative", "file", "https"}
        == {
            "relative"
            if not (path := value.path.path).startswith(("file:", "http"))
            else path.split(":")[0]
            for group in (collection.nodes, study.nodes, derived.root.nodes)
            for value in group
        },
    ),
    ("a document round-trips unchanged", round_trip_intact),
    (
        "a collection may hold a path instead of its nodes",
        reopened.node("derived").nodes is None,
    ),
    (
        "following it yields a collection of its own",
        type(nested).__name__ == "NgffCollection",
    ),
    (
        "a reference resolves across two documents",
        resolved.images[0].data.shape == segmentations["mr"].data.shape,
    ),
    ("an unknown node type survives", foreign.type == "acme:mesh"),
    ("the scene declares the registration", transform.type == "displacements"),
    (
        "applying it improves organ overlap",
        np.mean([s[1] for s in scores]) > np.mean([s[0] for s in scores]),
    ),
]
for label, ok in checks:
    print(f"  {'yes' if ok else 'NO ':3s}  {label}")
assert all(ok for _, ok in checks)
```

## What is still open

RFC-8 is at status D1. Its own text lists most of what is missing, and it is
worth keeping that apart from what it does not mention.

The draft says three things about itself. Redundant metadata can go out of sync,
and keeping it consistent is left to implementations. Assembling a collection
costs one metadata request per node, and a future RFC may add a consolidated
form to cut the round trips. Object store paths and a vocabulary for
relationships between nodes are both listed as future possibilities. So anyone
waiting for S3 URLs, or for a way to say that one image is derived from another,
is waiting on work the authors have already scoped.

Two gaps are not on those lists.

**A reference cannot say what it expected to find.** A `Reference` is an id and
a path. A `Path` is a type and a location. Neither carries a checksum, a version
or a content hash. The security section covers whether a client should follow a
path at all, and says nothing about whether what is at the end of it is still
what was annotated. That matters most for the arrangement above, which is the
draft's own third user story: an output collection that links back to an input
image it does not own. The label maps here annotate an archive we do not
control, and if that archive is replaced they will quietly describe something
else. Content addressing already exists on the storage side. A path has no way
to write it down.

**Converting existing HCS metadata loses fields.** The draft says some
information in the current HCS layout is not yet represented, without saying
which. Comparing the schemas: a v0.4 acquisition carries a description, a start
time, an end time and a maximum field count, while an RFC-8 acquisition carries
an id and a name. The plate loses `field_count`. That is the provenance of when
a screen was run, and a migration today would drop it in silence.

One more thing is a requirement on readers rather than a gap in the format, and
this notebook does not meet it either. The security section says an
implementation must not follow paths that escape a sandbox without the user
agreeing, and should restrict which hosts it will fetch. Opening a collection
from a source you do not trust will today follow whatever paths it holds,
including absolute local paths and addresses inside your own network. Everything
opened above came from a place we chose, which is the only reason that was safe.
